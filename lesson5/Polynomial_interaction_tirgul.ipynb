    <ul style="margin:0 18px 0 0; padding:0;">
      <li><b>משוואה ופרמטרים:</b>  \( y_i = \beta_0 + \beta_1 x_i + \varepsilon_i \) — שינוי של יחידה ב-\(x\) משנה את תוחלת \(y\) ב-\(\beta_1\).</li>

      <li><b>הנחות המודל:</b> ליניאריות, תוחלת שגיאה 0, הומוסקדסטיות, אי-תלות, נורמליות (להסקה).</li>
      <li><b>מבחני השערות:</b> t לכל \(\beta_j\), ו-F להשוואת מודלים; שימוש ב-p-value ורווחי סמך (הסקה מתוך הסיכום מודל).</li>
    </ul>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge, RidgeCV
from sklearn.metrics import mean_squared_error, r2_score

import statsmodels.api as sm

rng = np.random.default_rng(42)
plt.rcParams['figure.figsize'] = (7,4)
```

### **מטלה א.1 :**  <br>
### הגדירו 250 דגימות וצרו נתונים סינתטיים לפי
### $$ y = 3 + 2x - 0.5x^2 + \varepsilon, \quad x\sim \mathcal{U}[-3,3],\ \varepsilon\sim \mathcal{N}(0,0.6^2).$$

### חלקו ל־Train/Test, התאימו מודל פולינומי בדרגות $d\in\{1,2,5\}$, והשוו **MSE** ו־\(R^2\) ב־Train וב־Test.

### ציירו את העקומות המותאמות מול הנתונים.

  </div>

<div dir="rtl"> 
    
### פתרון א.1


```python
# פתרון א.1
n = 250
x = rng.uniform(-3, 3, size=n)
eps = rng.normal(0, 0.6, size=n)
y = 3 + 2*x - 0.5*(x**2) + eps

X = x.reshape(-1,1)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
res = []
for d in [1,2,5]:
    pipe = Pipeline([('poly', PolynomialFeatures(degree=d, include_bias=False)),
                     ('lin', LinearRegression())])
    pipe.fit(X_tr, y_tr)
    yhat_tr = pipe.predict(X_tr)
    yhat_te = pipe.predict(X_te)
    res.append({'degree': d,
                'train_MSE': mean_squared_error(y_tr, yhat_tr),
                'test_MSE':  mean_squared_error(y_te, yhat_te),
                'train_R2':  r2_score(y_tr, yhat_tr),
                'test_R2':   r2_score(y_te, yhat_te)})
    xs = np.linspace(-3.2, 3.2, 400).reshape(-1,1)
    ys = pipe.predict(xs)
    plt.figure()
    plt.scatter(X_tr, y_tr, s=12, label='Train')
    plt.scatter(X_te, y_te, s=12, label='Test')
    plt.plot(xs, ys, label=f'degree {d}')
    plt.title(f" Polynom degree {d}")
    plt.xlabel('x'); plt.ylabel('y'); plt.legend(); plt.show()

pd.DataFrame(res)
```

### **מטלה ב.1 :** 

### דגמו 700 ערכים וצרו נתונים לפי
### $$ y = 4 + 1.5x_1 + 2x_2 + 3x_1x_2 + \varepsilon, \quad x_1,x_2\sim \mathcal{U}[-2,2]. $$

### התאימו מודל פולינומי בדרגה 2 עם `PolynomialFeatures(degree=2, include_bias=False)` ו־`LinearRegression`.  

### הציגו MSE/$R^2$, שלפו את שמות הפיצ'רים בעזרת `get_feature_names_out`, והציגו טבלת מקדמים כולל $x_1,x_2$.  

###  פרשו את מקדם האינטראקציה.
  </div>

# יצירת הדגימה

```python
import numpy as np

np.random.seed(42)          # לרפרודוציביליות
n = 700

# x1, x2 ~ U[-2,2]
x1 = np.random.uniform(-2, 2, size=n)
x2 = np.random.uniform(-2, 2, size=n)

# רעש Gaussי
eps = np.random.normal(0, 1, size=n)

# y לפי המודל
y = 4 + 1.5*x1 + 2*x2 + 3*x1*x2 + eps

X = np.column_stack([x1, x2])   # מטריצת פיצ'רים מקורית
```

# 2. PolynomialFeatures + LinearRegression (דרגה 2)

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# פולינום דרגה 2 בלי עמודת bias (האינטרספט יטופל ע״י המודל)
poly = PolynomialFeatures(degree=2, include_bias=False)

X_poly = poly.fit_transform(X)

# שמות הפיצ'רים אחרי ההרחבה
feature_names = poly.get_feature_names_out(['x1', 'x2'])
print(feature_names)
```

# פלט שמות הפיצ'רים:

```python
['x1' 'x2' 'x1^2' 'x1 x2' 'x2^2']
```

# עכשיו נתאים מודל רגרסיה ליניארית לפיצ'רים המורחבים:

```python
model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)

r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)

print("R^2:", r2)
print("MSE:", mse)
print("intercept:", model.intercept_)
print("coefficients:", model.coef_)
```

# 3. טבלת מקדמים מסודרת

```python
import pandas as pd

coef_table = pd.DataFrame({
    'feature': ['intercept'] + list(feature_names),
    'coef':    [model.intercept_] + list(model.coef_)
})

coef_table
```

### פתרון ב.1

```python
# פתרון ב.1
n = 700
x1 = rng.uniform(-2, 2, size=n)
x2 = rng.uniform(-2, 2, size=n)
eps = rng.normal(0, 0.6, size=n)
y = 4 + 1.5*x1 + 2.0*x2 + 3.0*(x1*x2) + eps

X = np.column_stack([x1, x2])
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=1)

poly = PolynomialFeatures(degree=2, include_bias=False)
Xtr = poly.fit_transform(X_tr)
Xte = poly.transform(X_te)

lin = LinearRegression().fit(Xtr, y_tr)
yhat_tr, yhat_te = lin.predict(Xtr), lin.predict(Xte)

print('Train MSE/R2:', mean_squared_error(y_tr, yhat_tr), r2_score(y_tr, yhat_tr))
print('Test  MSE/R2:', mean_squared_error(y_te, yhat_te), r2_score(y_te, yhat_te))

feat_names = poly.get_feature_names_out(['x1','x2'])
coef_df = pd.DataFrame({'feature': ['intercept'] + list(feat_names),
                        'coef': [lin.intercept_] + list(lin.coef_)})
coef_df
```

<h4 style="margin:4px 0 14px 0; font-size:20px; font-weight:700; color:#004c70;">     
הסבר אינטראקציה: <br> 
<br>
 כאשר מופיע איבר $x_1x_2$, ההשפעה השולית של $x_1$ על $y$ תלויה ב־$x_2$:  <br>
 $$ \partial y/\partial x_1 = \beta_{x_1} + \beta_{x_1x_2}\,x_2 $$<br>
<br>
 באותו אופן ההשפעה השולית של $x_2$ תלויה ב- $x_1$<br>
 $$ \partial y/\partial x_2 = \beta_{x_2} + \beta_{x_1x_2}\,x_1 $$
</h4>

### 1. $ y = 2 + 2x_1 + 10\,promo + 5\,x_1\cdot promo + \varepsilon $ 

### 2.   התאימו מודל ליניארי בעזרת `statsmodels.OLS` על הפיצ'רים `[1, x1, promo, x1*promo]`, הציגו טבלת סיכום ופרשו את המקדמים (כולל ערכי-\(p\)).

### 3. נא להציג ויזואלית את הנתונים עבור promo = 0  ועבור promo = 1 ואת ההתאמה הליניארית שלהם בנפרד. 


### 4. בצעו שתי משוואת רגרסיה פשוטות שחוזים את y בעזרת x1 : <br> פעם עבור הנתונים של promo = 0 <br> ופעם השנייה עבור promo = 1. <br> נא להסביר את הערכים שקיבלתם במקדמים בעזרת המקדמים שקיבלתם במודל עם האינטראקציה

```python
# פתרון ד.1
n = 600
x1 = rng.uniform(-2, 2, size=n)
promo = rng.integers(0, 2, size=n)
eps = rng.normal(0, 0.7, size=n)
y = 2 + 2*x1 + 10*promo + 5*(x1*promo) + eps

X = pd.DataFrame({'const': 1.0, 'x1': x1, 'promo': promo, 'x1_x_promo': x1*promo})
df = pd.DataFrame({'const': 1.0, 'x1': x1, 'promo': promo, 'x1_x_promo': x1*promo, 'y':y})

X
# פרשנות קצרה:
# 'promo' משנה את האינטרספט, ו-'x1_x_promo' משנה את השיפוע של x1 כאשר promo=1.
```
