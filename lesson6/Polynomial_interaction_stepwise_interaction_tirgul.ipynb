      <li><b>מבחני השערות:</b> t לכל \(\beta_j\), ו-F להשוואת מודלים; שימוש ב-p-value ורווחי סמך (הסקה מתוך הסיכום מודל).</li>

      נגדיר:

<ul>
      <li><b>$x_1$ = שעות לימוד</b></li>
      <li><b>$x_2$ = שעות שינה</b></li>
      <li><b>$y$ = ציון במבחן</b></li>
    </ul>

<br>

<div style="font-weight:700; font-size:120%; margin-bottom:4px;">
      1. מודל בלי אינטראקציה – "כל אחד עובד לבד"
    </div>
      נבחר מודל פשוט ללא אינטראקציה:
      $y = 20 + 5x_1 + 2x_2$.
    </p>
      במודל הזה כל משתנה "עובד לבד":
    </p>

<ul>
      <li>כל שעה לימוד ($x_1$) מוסיפה בערך עוד 5 נקודות לציון $y$ – בלי קשר לכמה ישנת.</li>
      <li>כל שעה שינה ($x_2$) מוסיפה בערך עוד 2 נקודות לציון $y$ – בלי קשר לכמה למדת.</li>
    </ul>
      אם נצייר גרפים של $y$ לפי $x_1$ לשתי רמות שונות של שינה (למשל $x_2 = 4$ ו־$x_2 = 8$):
    </p>

<ul>
      <li>בשני המקרים השיפוע לפי $x_1$ הוא אותו דבר (5 נקודות לכל שעה לימוד).</li>
      <li>הקווים יהיו מקבילים – רק ה־intercept זז למעלה כשיש יותר שעות שינה.</li>
    </ul>
      כלומר: המודל אומר שהתרומה של שעות לימוד לציון <b>אינה תלויה</b> בשעות השינה, והתרומה של שעות שינה לציון <b>אינה תלויה</b> בשעות הלימוד.
    </p>

<br>

<div style="font-weight:700; font-size:120%; margin-bottom:4px;">
      2. מודל עם אינטראקציה – לשילוב יש אפקט מיוחד
    </div>
      עכשיו נוסיף איבר אינטראקציה בין שעות לימוד לשעות שינה, למשל:
      $y = 60 - 4x_1 - 2.5x_2 + 1.5x_1 x_2$.
    </p>
      כאן יש לנו לא רק את ההשפעה של $x_1$ ושל $x_2$ בנפרד, אלא גם איבר שמייצג את
      <b>השילוב ביניהם</b> – $x_1 x_2$.
    </p>
      ההשפעה השולית של שעות לימוד על הציון היא:
      $\frac{\partial y}{\partial x_1} = -4 + 1.5x_2$.
    </p>

<ul>
      <li>אם ישנת מעט, נניח $x_2 = 4$ שעות, אז $\frac{\partial y}{\partial x_1} = -4 + 1.5 \cdot 4 = 2$ –
          כל שעה לימוד מוסיפה בערך 2 נקודות בלבד.</li>
      <li>אם ישנת הרבה, נניח $x_2 = 8$ שעות, אז $\frac{\partial y}{\partial x_1} = -4 + 1.5 \cdot 8 = 8$ –
          כל שעה לימוד מוסיפה בערך 8 נקודות, הרבה יותר משתלם ללמוד כשישנת טוב.</li>
    </ul>
      אם נצייר גרפים של $y$ לפי $x_1$ לשתי רמות שונות של שינה:
    </p>

<ul>
      <li>עבור $x_2 = 4$ נקבל קו עם שיפוע מתון (בערך 2 נקודות לכל שעה לימוד).</li>
      <li>עבור $x_2 = 8$ נקבל קו עם שיפוע חד יותר (בערך 8 נקודות לכל שעה לימוד).</li>
      <li>הקווים כבר <b>לא מקבילים</b> – יש להם שיפועים שונים.</li>
    </ul>
      זה בדיוק הרעיון של אינטראקציה: 
      <b>התועלת מעוד שעה לימוד תלויה בכמה ישנת,
      והתועלת מעוד שעה שינה תלויה בכמה למדת</b>.
      השילוב ביניהם יוצר אפקט מיוחד מעבר לסכום ההשפעות הנפרדות.
    </p>

  </div>
</div>


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures

np.random.seed(42)

# ----- No interaction -----
def model_no_interaction(x1, x2):
    return 20 + 5*x1 + 2*x2

# ----- With interaction -----
def model_with_interaction(x1, x2):
    return 60 - 4*x1 - 2.5*x2 + 1.5*x1*x2

hours_study = np.repeat(np.arange(0, 5), 2)       # 0..4
hours_sleep = np.tile(np.array([4, 8]), 5)        # 4 or 8

df = pd.DataFrame({
    "hours_study": hours_study,
    "hours_sleep": hours_sleep
})

# Generate grades + small noise
df["grade_no_int"]  = round(model_no_interaction(df["hours_study"], df["hours_sleep"])  + np.random.normal(0, 2, len(df)), 1)
df["grade_int"]     = round(model_with_interaction(df["hours_study"], df["hours_sleep"]) + np.random.normal(0, 2, len(df)),1)

df
```

```python
sns.lmplot(
    data=df,
    x="hours_study", y="grade_no_int",
    hue="hours_sleep",
    markers=["o", "s"],
    ci=None
)
plt.title("No interaction: parallel lines");
```

```python
sns.lmplot(
    data=df,
    x="hours_study", y="grade_int",
    hue="hours_sleep",
    markers=["o", "s"],
    ci=None
)
plt.title("With interaction: different slopes");
```

```python
# Function to highlight maximum in red
def highlight_max_red(s):
    is_max = s == s.max()
    return ['color: red' if v else '' for v in is_max]

# Function to highlight minimum in red
def highlight_min_red(s):
    is_min = s == s.min()
    return ['color: red' if v else '' for v in is_min]
```

### $$ f(X) =  \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \beta_3 X_1 X_2 + \beta_4 X_1^2 + \beta_5 X_2^2 + \beta_6 X_1 X_2^2 + \beta_7 X_1^2 X_2  $$
  
## ייתכן שהאינטראקציות אינן ליניאריות, או לא רק ליניאריות, לכן נרצה לבדוק את כל האופציות. 
### בדוגמא הבאה נרצה לחזות שווי בית לפי קרבה לרכבת בק"מ קרבה למרכז העיר וגובה זיהום אוויר (מספר מ-1 עד 100)

  </div>

```python
# Section 2: Synthetic Data Generation
np.random.seed(0)
n = 300
train_dist = np.random.uniform(0, 20, n)
downtown_dist = np.random.uniform(0, 2, n)
air_pollution = np.random.randint(0, 100, n)

# Nonlinear relationship with interactions
revenue = (7500 + 10.8 * train_dist + 4.5 * downtown_dist - 0.2 * air_pollution +
    0.002 * train_dist * downtown_dist + 0.0003 * downtown_dist**2 - 0.5 * air_pollution**2 +  0.1 * train_dist**3 +  1.1 * train_dist**2 * downtown_dist + 
    np.random.normal(0, 150, n) )

X = np.column_stack((train_dist, downtown_dist, air_pollution))

X_data = pd.DataFrame(X, columns =  ['train_dist', 'downtown_dist', 'air_pollution'])

y = revenue.reshape(-1, 1)

X_train, X_test, y_train, y_test = train_test_split(X_data, y, test_size=0.3, random_state=1)

X_train.head()
```

## ייתכן שהאינטראקציות אינן ליניאריות, או לא רק ליניאריות, לכן נרצה לבדוק את כל האופציות. 

### כאשר מתחשבים באינטראקציות מסדר גדול יותר יש הרבה יותר משתנים.

### נראה עבור שלושה משתנים מסבירים לכמה זה מסתכם אם עושים רגרסיה פולינומית עם אינטראקציות מסדר n

```python
num_fetures = []
for degree in range(1,11):
    poly = PolynomialFeatures(degree=degree, include_bias=True)
    X_train_poly = poly.fit_transform(X_train)
    feature_names = poly.get_feature_names_out(X_train.columns)
    df_X_train = pd.DataFrame(X_train_poly, columns= feature_names)
    print('Degree, ', degree, 'Number of ploynomial features',  df_X_train.shape[1])
    num_fetures.append(df_X_train.shape[1])
```

</ul>

<br>
זה ממחיש את הבעיה של <b>"פיצוץ במספר הפיצ'רים"</b> כשעובדים עם רגרסיה פולינומית ואינטראקציות מסדר גבוה:

```python
# Section 3: Fit and Evaluate Polynomial Models
def fit_and_evaluate_poly(X_train, y_train, X_test, y_test, degree):

    poly = PolynomialFeatures(degree=degree, include_bias=False)
    
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)

    model = sm.OLS(y_train, sm.add_constant(X_train_poly)).fit()
    y_pred = model.predict(sm.add_constant(X_test_poly))

    rss = np.sum((y_test - y_pred) ** 2)
    aic = model.aic
    bic = model.bic
    r2_adj = model.rsquared_adj

    return model, y_pred, rss, aic, bic, r2_adj, poly.get_feature_names_out()
```

```python
# Section 4: Model Comparison
results = []
models = {}

for degree in range(1, 11):
    model, y_pred, rss, aic, bic, r2_adj, features = fit_and_evaluate_poly(X_train, y_train, X_test, y_test, degree)
    results.append({
        'Degree': degree,
        'AIC': aic,
        'BIC': bic,
        'Adjusted R^2': r2_adj
    })
    models[degree] = (model, y_pred, features)

results_df = pd.DataFrame(results)
```

    Stepwise Selection — בחירת משתנים אוטומטית לרגרסיה
  </h2>

  <div style="height:1px; background:#cbeafb; margin:4px 0 18px 0;"></div>
  
###  לא נרצה לבחור את כל המשתנים מפולינום מסדר כלשהו אלא רק לבחור רק חלק מהם.

### ישנן שיטות סטטיסטיות היורסיטיות לבחור אילו משתנים נכנסנים למודל ואילו לא. 
  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
    מה זה שיטה היוריסטית?
  </div>    

  <div style="font-size:15px; color:#0a5773;">
    שיטה היוריסטית היא שיטה שאינה מבטיחה למצוא את הפתרון האופטימלי בכל מחיר,
    אבל לרוב מאוד יעילה ומגיעה לפתרון <b>טוב וקרוב לאופטימלי</b> בזמן סביר.
    Stepwise Selection היא דוגמה מצוינת לשיטה כזו: לא סורקת את כל המודלים האפשריים,
    אלא בונה/מצמצמת את המודל צעד־אחר־צעד לפי קריטריון סטטיסטי.
  </p>
  <br>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
    מה זה Stepwise Selection?
  </div>

<div style="font-size:15px; color:#0a5773;">
    Stepwise Selection היא שיטה אוטומטית לבחירת משתנים (features) במודל רגרסיה.
    המטרה: למצוא <b>תת־קבוצה טובה של משתנים</b> שמסבירה את $y$ היטב,
    בלי להכניס למודל יותר מדי משתנים מיותרים.
  </p>

  <br>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:0 0 6px 0;">
    1. Forward Selection
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li>מתחילים ממודל <b>ריק</b> (רק אינטרספט, בלי משתנים מסבירים).</li>
    <li>בכל צעד בוחרים את המשתנה שהכי "משפר" את המודל (למשל, בעל ה־p-value הנמוך ביותר, או זה שמקטין הכי הרבה AIC/BIC).</li>
    <li>מוסיפים אותו למודל <b>רק אם</b> הוא עומד בקריטריון (למשל $p < 0.05$).</li>
    <li>ממשיכים להוסיף משתנים צעד־אחר־צעד, עד שאין שיפור משמעותי.</li>
  </ul>

  <br>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:0 0 6px 0;">
    2. Backward Elimination
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li>מתחילים ממודל שכולל את <b>כל</b> המשתנים המועמדים.</li>
    <li>בכל צעד מאתרים את המשתנה שהכי פחות "תורם" (למשל, בעל ה־p-value הגבוה ביותר).</li>
    <li>אם ה־p-value שלו גדול מסף (למשל $p > 0.05$) – מסירים אותו מהמודל.</li>
    <li>ממשיכים להסיר צעד־אחר־צעד, עד שכל המשתנים שנשארו מובהקים לפי הקריטריון.</li>
  </ul>

  <br>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:0 0 6px 0;">
    3. Stepwise Selection (שילוב של השניים)
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li>מתחילים כמו ב־Forward (או כמו ב־Backward), אבל בכל צעד:</li>
    <li>אפשר גם <b>להוסיף</b> משתנה חדש שלא במודל, וגם <b>להסיר</b> משתנה שכבר בפנים אם הוא כבר לא מובהק.</li>
    <li>השיטה מנסה לשמור על מודל "מאוזן" – לא ארוך מדי, לא קצר מדי.</li>
    <li>הרעיון: לא להיתקע במודל גרוע רק בגלל שהתחלנו מנקודת מוצא מסוימת.</li>
  </ul>

  <br>
  
<div style="font-size:15px; color:#0a5773;">
    בפועל, אפשר לבצע Stepwise לפי קריטריונים סטטיסטיים שונים:
    p-value, AIC, BIC, או קריטריונים אחרים. הרעיון זהה:
    <b>להוסיף/להסיר משתנים בצורה חכמה והיוריסטית</b>,
    במקום להריץ את כל הקומבינציות האפשריות (מה שלא ריאלי כשיש הרבה פיצ'רים).
  </p>

  </div>
</div>


```python
import statsmodels.api as sm

def stepwise_selection(X_train, y, degree, initial_list=[], threshold_in=0.05, threshold_out=0.05, verbose=True):

    poly = PolynomialFeatures(degree=degree, include_bias=False)
    X_train_poly = poly.fit_transform(X_train)
    feature_names = poly.get_feature_names_out(['train_dist', 'downtown_dist', 'air_pollution'])
    
    X = pd.DataFrame(X_train_poly, columns=feature_names)
     
    included = list(initial_list)
    while True:
        changed = False
        # forward step
        excluded = list(set(X.columns) - set(included))
        new_pvals = pd.Series(index=excluded, dtype=float)
        for new_column in excluded:
            model = sm.OLS(y, sm.add_constant(X[included + [new_column]])).fit()
            new_pvals[new_column] = model.pvalues[new_column]
        best_pval = new_pvals.min()
        if best_pval < threshold_in:
            best_feature = new_pvals.idxmin()
            included.append(best_feature)
            changed = True
            if verbose:
                print(f'Add  {best_feature} with p-value {best_pval:.6}')

        # backward step
        model = sm.OLS(y, sm.add_constant(X[included])).fit()
        pvals = model.pvalues.iloc[1:]  # exclude intercept
        worst_pval = pvals.max()
        if worst_pval > threshold_out:
            worst_feature = pvals.idxmax()
            included.remove(worst_feature)
            changed = True
            if verbose:
                print(f'Drop {worst_feature} with p-value {worst_pval:.6}')
        if not changed:
            break

    return included
```

  <ul style="font-size:15px; color:#0a5773;">
    הפונקציה <code>stepwise_selection</code> מבצעת בחירה אוטומטית של משתנים במודל רגרסיה
    באמצעות שיטת Stepwise על גבי פיצ'רים פולינומיאליים (כולל אינטראקציות) שנבנים מתוך

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:10px 0 6px 0;">
    1. יצירת פיצ'רים פולינומיאליים
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li>הפונקציה בונה אובייקט <code>PolynomialFeatures</code> בדרגה שנבחרה (<code>degree</code>).</li>
    <li>הפרמטר <code>include_bias=False</code> אומר שלא מוסיפים עמודת 1, כי האינטרספט יתווסף אחר כך עם <code>sm.add_constant</code>.</li>
    <li>הקריאה <code>fit_transform(X_train)</code> יוצרת את כל האיברים הפולינומיאליים והאינטראקציות.</li>
    <li>הקריאה <code>get_feature_names_out</code> מחזירה שמות ברורים לכל פיצ'ר, כגון:
        <code>"train_dist^2"</code>, <code>"train_dist air_pollution"</code> ועוד.</li>
    <li>בונים <code>DataFrame</code> בשם <code>X</code> שמכיל את כל הפיצ'רים הפולינומיאליים.</li>
  </ul>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:10px 0 6px 0;">
    2. אתחול רשימת המשתנים והלולאה הראשית
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li><code>included</code> היא רשימת הפיצ'רים שנמצאים כרגע במודל (מתחילים מ־<code>initial_list</code>, לרוב ריק).</li>
    <li>בתוך <code>while True</code> מתבצע כל פעם צעד של Forward וצעד של Backward.</li>
    <li>המשתנה <code>changed</code> מסמן אם הוספנו או הסרנו משתנה בסיבוב הנוכחי.</li>
  </ul>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:10px 0 6px 0;">
    3. שלב Forward — הוספת משתנים
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li>מחשבים את רשימת המשתנים שאינם במודל: <code>excluded</code>.</li>
    <li>עבור כל משתנה מועמד מריצים מודל OLS עם כל המשתנים שכבר בפנים + המועמד החדש.</li>
    <li>שומרים את ה־p-value של כל מועמד בסדרה <code>new_pvals</code>.</li>
    <li>מוצאים את המועמד עם ה־p-value הנמוך ביותר:
        אם הוא קטן מ־<code>threshold_in</code> (למשל 0.05) – מוסיפים אותו ל־<code>included</code>.</li>
    <li>אם <code>verbose=True</code>, מודפס מסר בסגנון: "Add X with p-value ...".</li>
  </ul>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:10px 0 6px 0;">
    4. שלב Backward — הסרת משתנים
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li>מריצים מודל OLS על כל המשתנים שנמצאים כרגע ב־<code>included</code>.</li>
    <li>מוציאים את ה־p-values של כל הפיצ'רים (למעט האינטרספט, שנמצא בעמודה הראשונה).</li>
    <li>מוצאים את המשתנה עם ה־p-value הגבוה ביותר.</li>
    <li>אם ה־p-value שלו גדול מ־<code>threshold_out</code> – מסירים אותו מהמודל.</li>
    <li>גם כאן, אם <code>verbose=True</code>, מודפס מסר: "Drop X with p-value ...".</li>
  </ul>

  <div style="font-size:16px; color:#0a5773; font-weight:600; margin:10px 0 6px 0;">
    5. תנאי העצירה והפלט
  </div>

  <ul style="font-size:15px; color:#0a5773;">
    <li>אם בסיבוב מסוים לא הוסף ואף לא הוסר אף משתנה (<code>changed</code> נשאר False) — הלולאה נעצרת.</li>
    <li>בסיום התהליך הפונקציה מחזירה את הרשימה <code>included</code> — זוהי קבוצת המשתנים שנבחרו על ידי Stepwise עבור הפולינום בדרגה שנבחרה.</li>
  </ul>
