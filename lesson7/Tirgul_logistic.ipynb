  <h2 style="margin:4px 0 14px 0; font-size:48px; font-weight:700; color:#004c70;">
    רגרסיה לוגיסטית
  </h2>

  <div style="height:1px; background:#cbeafb; margin:4px 0 18px 0;"></div>

  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
ניקח צעד אחורה ונדבר על רגרסיה לינארית:<br>
ברגרסיה לינארית עבדנו עם משתנה מטרה רציף והתאמנו קו ישר שמתאר את הקשר בין המשתנה המסביר למוסבר.<br>
חישבנו באמצעות $R^2$ כמה טוב הקו מסביר את השונות ב-$Y$. <br>
בדקנו מובהקות כדי לקבוע האם הקשר בין $X$ ל-$Y$ מובהק סטטיסטית.<br>
לבסוף השתמשנו בקו הישר כדי לנבא את $Y$ בהינתן ערכים חדשים של $X$.<br>
<br>
רגרסיה לוגיסטית דומה לרגרסיה לינארית – אבל כאן משתנה המטרה $Y$ הוא בינארי (למשל 0/1, אמת/שקר, קבוצה A / קבוצה B),<br>
והמטרה שלנו היא לנבא <u>הסתברות</u> ש-$Y=1$ ולא מספר רציף כללי.<br>


<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<span style="font-size:26px; font-weight:700;">1. מה אנחנו מנסים לחזות ברגרסיה לוגיסטית?</span><br><br>
נניח ש-$Y \in \{0,1\}$ (למשל: $Y=1$ = "עבר את המבחן", $Y=0$ = "לא עבר").<br>
יש לנו וקטור מסבירים $X = (x_1, x_2, \dots, x_p)$.<br>
המטרה של המודל היא לחזות את ההסתברות:<br>
$P(Y=1 \mid X=x)$<br>
כלומר: מה הסיכוי ש-$Y=1$ עבור סט מאפיינים $x$.<br>


<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<span style="font-size:26px; font-weight:700;">2. הצגת המשוואה – פונקציית ה-sigmoid</span><br><br>
הרעיון המרכזי:<br>
קודם מחשבים ציון לינארי (כמו ברגרסיה רגילה):<br>
$z = \beta_0 + \beta_1 x_1 + \dots + \beta_p x_p$<br>
ואז "מעבירים" אותו דרך פונקציה לוגיסטית (sigmoid) שמחזירה ערך בין 0 ל-1:<br>
$$\sigma(z) = \dfrac{1}{1 + e^{-z}}$$<br>
<br>
לכן, המודל הלוגיסטי כותבים כך:<br>
$$P(Y=1 \mid X=x) = \sigma(\beta_0 + \beta_1 x_1 + \dots + \beta_p x_p)$$<br>

<br>
<b>אינטואיציה לפונקציית sigmoid:</b><br>
- כאשר $z \to -\infty$ נקבל $\sigma(z) \approx 0$ (הסתברות נמוכה מאוד).<br>
- כאשר $z \to +\infty$ נקבל $\sigma(z) \approx 1$ (הסתברות גבוהה מאוד).<br>
- כאשר $z = 0$ נקבל $\sigma(0) = 0.5$.<br>
כלומר, $\sigma(z)$ היא פונקציה חלקה בצורת S שממירה כל מספר ממשי להסתברות חוקית בין 0 ל-1.<br>


<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<span style="font-size:26px; font-weight:700;">3. למה לא להשתמש ברגרסיה לינארית רגילה להסתברויות?</span><br><br>
אם ננסה מודל לינארי כמו:<br>
$$\hat{Y} = \beta_0 + \beta_1 x_1 + \dots + \beta_p x_p$$<br>
כאשר $Y$ הוא 0/1:<br>
- אפשר לקבל "הסתברויות" שליליות או גדולות מ-1 – לא הגיוני כהסתברות.<br>
- ההנחות על השגיאה (נורמליות, שונות קבועה) כבר לא נכונות כש-$Y$ בינארי.<br>
לכן, אנחנו לא מאמנים מודל בינארי עם $OLS$, אלא משתמשים ברגרסיה לוגיסטית עם פונקציית $sigmoid$ ואמידה אחרת של הפרמטרים.<br>


<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<span style="font-size:26px; font-weight:700;">4. חישוב הפרמטרים – פונקציית נראות (Likelihood)</span><br><br>
נסמן $p_i = P(Y_i=1 \mid X_i)$ עבור תצפית מספר $i$.<br>
במודל הלוגיסטי: $p_i = \sigma(\beta_0 + \beta_1 x_{i1} + \dots + \beta_p x_{ip})$.<br>
מאחר ש-$Y_i \in \{0,1\}$, <br>
נניח ש-$Y_i \sim Bernoulli(p_i)$<br>
כלומר:<br>
$$P(Y_i \mid X_i) = p_i^{Y_i} (1 - p_i)^{1-Y_i}$$<br>
<br>
פונקציית הנראות (Likelihood) לכל המדגם היא מכפלת ההסתברויות:<br>
$$L(\beta) = \prod_{i=1}^n p_i^{Y_i} (1 - p_i)^{1-Y_i}$$<br>
בפועל עובדים עם לוג-נראות (log-likelihood):<br>
$$\ell(\beta) = \log L(\beta) = \sum_{i=1}^n \big[ Y_i \log(p_i) + (1 - Y_i)\log(1 - p_i) \big]$$<br>
<br>
עיקרון האמידה: בוחרים את $\hat{\beta}$ כך ש-$\ell(\beta)$ תהיה מקסימלית – זו שיטת <u>Maximum Likelihood</u>.<br>
אין כאן פתרון סגור ויפה כמו ב-OLS, ולכן האמידה נעשית עם אלגוריתמים נומריים (ניוטון-רפסון, Gradient Descent וכו') שממומשים עבורנו בספריות סטטיסטיות.<br>


<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<span style="font-size:26px; font-weight:700;">5. תחזית בעזרת רגרסיה לוגיסטית</span><br><br>
לאחר שאמדנו את הפרמטרים וקיבלנו $\hat{\beta}$, עבור תצפית חדשה $x^{(new)}$:<br>
1. מחשבים ציון לינארי:<br>
$z^{(new)} = \hat{\beta}_0 + \hat{\beta}_1 x^{(new)}_1 + \dots + \hat{\beta}_p x^{(new)}_p$<br>
2. ממירים להסתברות:<br>
$\hat{p} = \sigma(z^{(new)}) = \dfrac{1}{1 + e^{-z^{(new)}}}$<br>
3. אם רוצים החלטה בינארית (0/1), בוחרים סף (למשל 0.5):<br>
אם $\hat{p} \ge 0.5$ נחזה $Y=1$, ואם $\hat{p} < 0.5$ נחזה $Y=0$.<br>
<br>
חשוב: המודל הלוגיסטי בעצמו נותן לנו <u>הסתברות</u>. ההמרה למסקנה "0/1" היא שלב החלטה שמבוסס על סף שנבחר לפי ההקשר (עלות טעויות, רגישות, ספציפיות ועוד).<br>
    

<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<span style="font-size:26px; font-weight:700;">6. פונקציית logit והמשוואה הלינארית בלוג-אודס</span><br><br>
פונקציית $logit$ מוגדרת כך:<br>
$$logit(p) = \log\left(\dfrac{p}{1-p}\right)$$<br>
<br>
אם ניקח את $p = P(Y=1 \mid X)$ במודל הלוגיסטי, אפשר לכתוב את המודל בצורה הבאה:<br>
$logit\big(P(Y=1 \mid X)\big) = \beta_0 + \beta_1 x_1 + \dots + \beta_p x_p$<br>
כלומר, לא ההסתברות עצמה לינארית ב-$X$, אלא <u>הלוג-אודס</u> (log-odds) שלהם.<br>
<br>
הפונקציה $logit$ היא ההפך של ה-$sigmoid$:<br>
אם $p = \sigma(z)$ אז $z = logit(p)$.<br>
<br>
בשורה תחתונה :<br>
$logit$ לוקח הסתברות $p$ והופך אותו למספר כלשהו על כל ציר המספרים (יכול להיות חיובי, שלילי או 0).<br>
הוא מחשב את הלוג-אודס, כלומר : לוג טבעי של יחס הסיכוי להצלחה או כישלון.<br>
היתרון : אחרי $logit$ , הקשר בין $X$ לבין $logit(p)$ נהיה לינארי (אפשר להשתמש ברגרסיה), ואפשר לחזור חזרה להסתברות בעזרת פונקציית סיגמויד.

<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<span style="font-size:26px; font-weight:700;">7. המושגים odds, log-odds ו-Odds Ratio</span><br><br>
<b>Odds</b> (אודס) של הסתברות $p$ מוגדרים כיחס בין הסיכוי להצלחה לסיכוי לכישלון:<br>
$odds = \dfrac{p}{1-p}$<br>
<br>
דוגמאות:<br>
- אם $p = 0.8$ אז $odds = \dfrac{0.8}{0.2} = 4$ → "4 לטובת הצלחה על כל 1 נגד".<br>
- אם $p = 0.2$ אז $odds = \dfrac{0.2}{0.8} = 0.25$.<br>
<br>
<b>log-odds</b> (או $logit$):<br>
$log\text{-}odds = \log\left(\dfrac{p}{1-p}\right)$<br>
זה בדיוק הצד השמאלי במשוואת הרגרסיה הלוגיסטית:<br>
$log\left(\dfrac{P(Y=1 \mid X)}{1 - P(Y=1 \mid X)}\right) = \beta_0 + \beta_1 x_1 + \dots + \beta_p x_p$<br>

<br>
<b>פרשנות של $\beta_j$ – Odds Ratio</b><br>
נניח שהמודל הוא:<br>
$$logit\big(P(Y=1 \mid X)\big) = \beta_0 + \beta_1 x_1 + \dots + \beta_p x_p$$<br>
אם נגדיל את $x_j$ ביחידה אחת (כששאר המשתנים קבועים), הלוג-אודס יגדלו ב-$\beta_j$.<br>
בעולם ה-odds עצמם, זה אומר שה-odds מוכפלים ב-$e^{\beta_j}$.<br>
<br>
לכן $e^{\beta_j}$ נקרא <b>Odds Ratio</b> (יחס אודס) עבור עליה של יחידה אחת ב-$x_j$.<br>
דוגמה: אם $\beta_1 = 0.7$ אז $e^{0.7} \approx 2.0$.<br>
המשמעות: על כל עליה של יחידה אחת ב-$x_1$, ה-odds ל-$Y=1$ (למשל "עובר") מוכפלים בערך פי 2, בהנחה ששאר המשתנים לא משתנים.<br>

<div style="background:#e0f4ff; border:2px solid #0077b6; border-radius:12px; padding:20px 24px;
            direction:ltr; text-align:right;">
  <div dir="rtl" style="font-size:22px; color:#0a5773; font-weight:600; margin:0 0 10px 0;">
<b>לסיכום:</b> <br>
- רגרסיה לוגיסטית היא כלי בסיסי לסיווג בינארי, שבו מנבאים הסתברות $P(Y=1 \mid X)$.<br>

## פונקציה לוגיסטית
    
###     בשביל להבטיח שהמודל תמיד יתן ערך בין 0 ל-1 כיאה להסתברות, נשתמש באיזו פונקציה שנקראת logistic (בעולם של רשתות ניורונים קוראים לזה sigmoid). הפונקצייה נראית כך:
### $$ logistic(x) = \frac{1}{1+e^{-x}} $$
    
###    נראה ויזואלית איך שזה נראה: 



```python
def logistic(x):
    return 1/(1+np.exp(-x))

x_vals = np.linspace(-7.5, 7.5, 100)
plt.figure(figsize=(8,3))
plt.plot(x_vals, logistic(x_vals))
plt.xlabel('X', fontsize = 18)
plt.ylabel('Y',  fontsize = 18)
plt.plot(x_vals,0.5*np.ones(100), color = 'black', label = 'Y = 0.5')
plt.grid()
plt.title('Sigmoid function',  fontsize = 22)
plt.xlim(-7.5,7.5)
plt.legend()
plt.show()
```

### $$\frac{p}{1-p}$$

## אז איך משתמשים בפונקציה הלוגיסטית הזאת כדי לעשות משוואת רגרסיה?

<div dir="rtl">  בדוגמא למעלה הוצגה המשוואה הלוגיסטית עבור X, במקרה של רגרסיה לוגיסטית המשוואה תראה כך:
נרצה לחזות את ההסתברות לכשל, נסמן זאת על ידי $\hat{p}_{failure}$. ברגרסיה לוגיסטית גם שם יש משוואה ליניאירית. היא נראית כך:
    
## $$ x = b_0 +b_1 *temperature,   $$
    
    
## $$ \hat{p}_{failure} =  logistic(x) = logistic(b_0 +b_1 *temperature) = \frac{1}{1+e^{-(b_0 +b_1 *temperature)}} $$
    



<div dir="rtl">

##    גם כאן יש לנו 2 פרמטרים למצוא. אנחנו שואפים למצוא את הערכים בעזרת נתנוני הבעיה. ברגרסיה ליניארית מצאנו את הפרמרטים על ידי מזעור השגיאות הריבעויות. ברגרסיה לוגיסטית שיטה זו לא אפשרית ומשתמשים בשיטה אחרת שנקראת נראות מקסימלית (Maximum likelihood).  
##    ננסה להבין את הרעיון של נראות מקסימלית על ידי דוגמא. נניח כי אנחנו רוצים למצוא את ההסתברות ששחקן כדורסל, נקרא לו סטף קרי יקלע לסל בזריקה בודדת. יהי n מספק הזריקות ויהי x מספק הקליעות. ננסה לשערך את הסיכוי לקליעה.
##    נסה למצוא את פונקציית הנראות ונראה איך הוא עוזרת לנו לשערך את הפונקצייה. נסמן ב$p$ את ההסתברות לקליעה.

## $$ L(p) = p^x (1-p)^{n-x} $$


<div dir="rtl">   

###    מסתבר שהערך של p שממקסם את הפונקצייה הזאת הוא:
    
## $$ p =  \frac{x}{n} $$
    
    
### נניח כי קרי זרק 10 פעמים וקלע 7, מהו הסיכוי קליעה בניסיון בודד שימקסם את הנראות??
    
## $$ p^7 (1-p)^3 $$

    
### נציג את הפוקציית נראות כפונקצייה של p.

$$ L(p) = p^7 (1-p)^3 $$
$$ d\frac{L(p)}{dp} = ? $$

$$ L(p) = p^7 (1-p)^3 $$



$$ p = \frac{7}{10} $$

```python
ps = np.linspace(0,1,300)
likelihood = (ps**(7)) * ((1-ps)**3)

plt.figure(figsize=(8,3))
plt.plot(ps, likelihood)
plt.vlines(0.7, 0, (0.7**(7)) * ((1-0.7)**3), ls = 'solid', color = 'black')
plt.title('Maximum likelihood' ,fontsize = 18)
plt.xlabel('P', fontsize = 20)
plt.ylabel('Likelihood', fontsize = 20)
plt.show()
```
