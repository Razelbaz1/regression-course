```python
import numpy as np
import matplotlib.pyplot as plt

# נתונים פשוטים להמחשה
x = np.linspace(0, 10, 100)

# מודל ליניארי על y^2
y_squared_hat = 4 + 2*x  # זה קו ישר במרחב של y^2

# החזרה לסקייל המקורי
y_hat = np.sqrt(y_squared_hat)

# גרף 1 – עולם ההתמרה (y^2)
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(x, y_squared_hat, color='royalblue', linewidth=3)
plt.title("The world of transformation – y² = 4 + 2x", fontsize=14)
plt.xlabel("x")
plt.ylabel("y²", rotation=0, labelpad= 10)
plt.grid(True)
# גרף 2 – העולם המקורי (y = √(4+2x))
plt.subplot(1,2,2)
plt.plot(x, y_hat, color='darkorange', linewidth=3)
plt.title("The original world – y = √(4 + 2x)", fontsize=14)
plt.xlabel("x")
plt.ylabel("y", rotation=0, labelpad= 10)
plt.grid(True)

plt.suptitle("A straight line in y² space becomes a curve when returned to y space", fontsize=16, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
```

במחברת הזו נלמד רגרסיה פולינומית על נתונים סינטטיים מעולם הבטיחות.

<div dir="rtl"

הסיפור:
- יש לנו **מלגזה במחסן**.
- המשתנה המסביר: מהירות נסיעה של המלגזה בקמ״ש.
- המשתנה התלוי: מדד סיכון בטיחותי (Risk Index) – ככל שהמדד גבוה יותר, המערכת פחות בטוחה.

<div dir="rtl"

נושאים שנכסה:
1. רגרסיה ליניארית פשוטה ובדיקת לינאריות.
2. טרנספורמציה על משתנה התגובה (לוג של ה-Risk).
3. בניית מודל פולינומי (חזקות של X) ושמירה על מודל ליניארי במקדמים.
4. הדגמת Overfitting עם פולינומים מדרגות שונות ופיצול Train/Test.
5. שימוש במדדים מענישי־מורכבות (Adjusted R², AIC, BIC) לבחירת דרגת הפולינום.

<div dir="rtl"

הנתונים יהיו סינטטיים, כך שה״אמת״ הסטטיסטית היא **פונקציה פולינומית ממעלה רביעית** של המהירות.


```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

import statsmodels.api as sm

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

np.random.seed(42)  # for reproducibility
sns.set(style="whitegrid")
```

```python
# Number of observations
n = 30

# Forklift speed in km/h, realistic range in a warehouse (0 to ~16 km/h)
speed = np.linspace(0, 30, n)

# Normalize speed around 8 km/h to get values roughly in [-2, 2]
x_norm = (speed - 15) / 7.5

# True underlying safety risk (5th-degree polynomial in x_norm)
# This polynomial has both even and odd powers -> inflection points, non-parabolic shape
true_risk = (
    10
    + 0.5 * x_norm         # linear term
    - 1.5 * x_norm**2      # quadratic term
    + 0.7 * x_norm**3      # cubic term
    + 0.3 * x_norm**4      # 4th-degree term
    - 0.2 * x_norm**5      # 5th-degree term
)

# Add random noise to simulate measurement error / variability
noise = np.random.normal(loc=0, scale=0.5, size=n)

risk = true_risk + noise

# Build DataFrame
df = pd.DataFrame({
    "forklift_speed_kmh": speed,
    "safety_risk_index": risk,
    "true_risk_no_noise": true_risk
})
```

```python
# אם ipywidgets לא מותקן:
# !pip install ipywidgets

import numpy as np
import matplotlib.pyplot as plt

import ipywidgets as widgets
from ipywidgets import interact
```

```python
def adjusted_r2(R2, n, k):
    """
    Compute Adjusted R^2 given:
    R2 - regular R^2
    n  - number of observations
    k  - number of predictors (without intercept)
    """
    return 1 - ((1 - R2) * (n - 1) / (n - k - 1))
```

```python
def plot_adjusted_r2_vs_k(R2=0.91, n=30, max_k=15):
    """
    Plot Adjusted R^2 as a function of k, for fixed R2 and n.
    """
    # k must be < n - 1 to avoid division by zero in the formula
    k_values = np.arange(1, max_k + 1)
    k_values = k_values[k_values < n - 1]

    R2_adj_values = adjusted_r2(R2, n, k_values)

    plt.figure(figsize=(8, 5))
    plt.plot(k_values, R2_adj_values, marker='o')
    plt.ylim(0, 1)
    plt.xlabel("k (number of predictors)")
    plt.ylabel("Adjusted R²")
    plt.title(f"Adjusted R² as a function of k\n(R² = {R2:.2f}, n = {n})")
    plt.grid(True)
    plt.show()

    # Display current Adjusted R² for the largest k in the range
    if len(k_values) > 0:
        k_current = k_values[-1]
        R2_adj_current = adjusted_r2(R2, n, k_current)
        print(f"For k = {k_current}, n = {n}, R² = {R2:.2f} → Adjusted R² = {R2_adj_current:.4f}")


interact(
    plot_adjusted_r2_vs_k,
    R2=widgets.FloatSlider(
        min=0.1, max=0.99, step=0.01, value=0.918552,  # 5	0.918552 0.901584
        description='R²'
    ),
    n=widgets.IntSlider(
        min=5, max=200, step=1, value=30,
        description='n'
    ),
    max_k=widgets.IntSlider(
        min=1, max=20, step=1, value=10,
        description='max k'
    )
);
```

### $$ R^2_{\text{adj}} = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - k - 1} \right) $$


```python
def plot_adjusted_r2_vs_n(R2=0.8, k=3, n_max=200):
    """
    Plot Adjusted R^2 as a function of n, for fixed R2 and k.
    """
    # n must be > k + 1 (otherwise denominator n - k - 1 <= 0)
    n_values = np.arange(k + 2, n_max + 1)
    R2_adj_values = adjusted_r2(R2, n_values, k)

    plt.figure(figsize=(8, 5))
    plt.plot(n_values, R2_adj_values, marker='o')
    plt.ylim(0, 1)
    plt.xlabel("n (number of observations)")
    plt.ylabel("Adjusted R²")
    plt.title(f"Adjusted R² as a function of n\n(R² = {R2:.2f}, k = {k})")
    plt.grid(True)
    plt.show()

    # Display Adjusted R² for the largest n
    if len(n_values) > 0:
        n_current = n_values[-1]
        R2_adj_current = adjusted_r2(R2, n_current, k)
        print(f"For n = {n_current}, k = {k}, R² = {R2:.2f} → Adjusted R² = {R2_adj_current:.4f}")


interact(
    plot_adjusted_r2_vs_n,
    R2=widgets.FloatSlider(
        min=0.1, max=0.99, step=0.01, value=0.8,
        description='R²'
    ),
    k=widgets.IntSlider(
        min=1, max=15, step=1, value=3,
        description='k'
    ),
    n_max=widgets.IntSlider(
        min=20, max=500, step=5, value=200,
        description='max n'
    )
);
```
